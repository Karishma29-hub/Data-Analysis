```python
import pandas as pd

data = {
    "Grade": ["A", "B", "C", "A"]
}

df = pd.DataFrame(data)

result = df["Grade"].map({
    "A": "Excellent",
    "B": "Good",
    "C": "Average"
})

print(result)
```

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Pooja", "Monesh"],
    "Marks": [90, 85]
})

df2 = pd.DataFrame({
    "Name": ["Karishma", "Moni"],
    "Marks": [88, 92]
})

result = pd.concat([df1, df2])

print(result)
```

```python
import pandas as pd

df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Pooja", "Ravi", "Anu"]
})

df2 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [90, 85, 88]
})

result = pd.merge(df1, df2, on="ID")

print(result)
```
